**Theoretical Questions**

**Question 1**: What is Data Loading in Power BI and why is it considered the first step of analysis?

**solution:**


In Power BI, Data Loading is the process of importing raw data from various sources (like Excel, SQL databases, or cloud services) into the Power BI In-Memory engine (VertiPaq) so you can start working with it.

While it feels like the very first thing you do, it is technically the culmination of the Extract, Transform, Load (ETL) pipeline. You extract the data from its source, transform it in Power Query (cleaning it up), and finally load it into the data model.

**Why It is Considered the First Step of Analysis**

Even though data cleaning happens just before it, data loading is officially where your actual data analysis begins. Here is why it sets the stage for everything else:

 * Establishes the Data Model: You cannot analyze isolated data. Loading the data brings all your separate tables into a single environment where you can establish relationships between them (e.g., linking a Sales table to a Customers table).

 * Activates the In-Memory Performance: Power BI doesn't just hold the data; it compresses it heavily into RAM. This loading step is what allows you to look at millions of rows and get instant calculations when you click a visual.

* Enables DAX (Data Analysis Expressions): Before you can write formulas to calculate profit margins, year-over-year growth, or running totals, the data must be fully loaded into the model's schema so DAX can navigate the tables.

 * Defines data types: During the load phase, Power BI locks in whether a column is a number, text, or date. Accurate analysis is impossible if your system tries to average a column it thinks is text.

**Question 2**: Explain the difference between “Load” and “Transform Data” in Power BI.

**solution:**

Choose "Load" if:

* Your data comes from a highly structured corporate data warehouse that is already pristine.

* You are importing a simple, pre-cleaned Excel tracking sheet.

* You just want to quickly preview how the raw data looks inside the Power BI dashboard environment.

Choose "Transform Data" if:

* Your date column is being read as text, or numbers are formatted incorrectly.

* You need to remove empty rows, duplicates, or unnecessary columns to keep your file size small.

* You need to combine multiple files or tables (e.g., merging a 2025_Sales file and a 2026_Sales file).

**Question 3 :** What is a Fact Table and a Dimension Table? Give examples from the dataset.

**solution:**

1. **Fact Table**

A Fact Table is the central table in your data model that holds the quantitative, measurable data resulting from a specific business event or transaction. It is full of numbers that you want to aggregate (sum, average, count) and foreign keys that link it to other tables.

Characteristics: High row count (grows rapidly), mostly numeric columns, and contains repetitive ID keys.

Answers the question: "What happened, how much, and when?"

2. **Dimension Table**

A Dimension Table contains the descriptive, qualitative attributes (the context) surrounding that business event. These tables are used to slice, dice, filter, and group the data in your Fact Table.

Characteristics: Wide tables with lots of text columns, lower row count, and unique keys (no duplicate IDs in the primary key column).

Answers the question: "Who, where, what product, or which department was involved?"

**Question 4 :** Why is Star Schema preferred over Snowflake Schema in Power BI?


**solution:**

In Power BI, the Star Schema is the gold standard for data modeling. While both Star and Snowflake schemas organize data using Fact and Dimension tables, they handle Dimension tables very differently.

**Star Schema:** Keeps dimension tables completely de-normalized (flattened). Each dimension is a single table directly connected to the fact table, looking like a star.

**Snowflake Schema:** Fully normalizes dimension tables by splitting them into sub-tables (e.g., splitting a Product dimension into separate Product, Sub-Category, and Category tables) to eliminate redundancy, looking like a complex snowflake.

Even though Snowflake schema saves a tiny amount of storage space by reducing duplicate text, Star Schema is heavily preferred in Power BI for several critical reasons:

**1. Drastically Faster Performance (The VertiPaq Engine)**

Power BI uses a highly advanced, In-Memory columnar database engine called VertiPaq.VertiPaq is built to compress repetitive text data incredibly well. The storage space you "save" by split-loading tables in a Snowflake schema is completely negligible in Power BI.

Instead, Snowflake schemas force Power BI to evaluate complex, multi-hop table relationships at runtime. Every time a user clicks a visual, Power BI has to traverse multiple table joins (e.g., Category $\rightarrow$ Sub-Category $\rightarrow$ Product $\rightarrow$ Sales Fact). A Star Schema requires only a single direct relationship hop, which makes queries lightning-fast.

**2. Simpler and Cleaner DAX Code**

Writing DAX (Data Analysis Expressions) formulas becomes significantly harder in a Snowflake schema.

Functions like RELATED(), CALCULATE(), and time intelligence filters rely on clear, directional relationship paths. When dimensions are split across multiple tables, your relationship directions can get tangled, forcing you to write messy, complex DAX code to traverse the hierarchy or deal with performance-killing bidirectional filters.

**3. Better User Experience for Report Designers**

When you or your teammates build reports in Power BI, you use the Fields Pane to drag and drop columns into charts.

In a Star Schema, all product attributes (ProductName, SubCategory, Category) live in one neat Product folder.

In a Snowflake Schema, the designer has to hunt through three or four different folders just to build a simple product breakdown chart, which is frustrating and inefficient.

**4. Optimized Auto-Exist and Aggregations**

Power BI has internal optimization mechanisms like Auto-Exist (which optimizes filters applied to the same table) and Aggregations (which speed up massive datasets). These features work best when descriptive attributes reside in the same dimension table rather than being fragmented across a snowflake structure.

**Question 7 :** Create a new column to calculate Recovery Rate.

**solution:**

**Create a New Column to Calculate Recovery Rate**

To calculate the Recovery Rate while preventing errors (like division by zero), use the following formula.

In Power BI (DAX Calculated Column):

Click on New Column in your data/report view and enter:



```
Recovery_Rate =
DIVIDE(
    'coronavirus_2020_dataset'[Recovered_Cases],
    'coronavirus_2020_dataset'[Confirmed_Cases],
    0
)
```

Remember to format this new column as a Percentage (%) under Column Tools).

Mathematical Formula Applied:

$$\text{Recovery Rate} = \frac{\text{Recovered_Cases}}{\text{Confirmed_Cases}}$$


**Question 8**: Create a summarized table showing total confirmed cases by Country.

**solution:**

You can create this either as a visual table on your dashboard or as a separate calculated table in your model.

In Power BI (DAX Calculated Table):

Go to the Modeling tab, click New Table, and enter:



```
Country_Summary_Table =
SUMMARIZE(
    'coronavirus_2020_dataset',
    'coronavirus_2020_dataset'[Country],
    "Total Confirmed Cases", SUM('coronavirus_2020_dataset'[Confirmed_Cases])
)
```
